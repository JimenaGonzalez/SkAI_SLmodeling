```python
import numpy as np
import pandas as pd
import os
import matplotlib.pyplot as plt

from astropy.visualization import make_lupton_rgb
plt.style.use('dark_background')
```

```python
path = '/Users/jimenagonzalez/research/SkAI_research/SkAI_SLmodeling/Simulations/data/Tianao/' 
deflector_light = 'lens_light_sersic-sie_sigma_noise=3.9e-03/' #lens_light_'
deflector_mass = 'lens_mass_sersic-sie_sigma_noise=3.9e-03/' #lens_mass_'
source_light = 'source_light_sersic-sie_sigma_noise=3.9e-03/' #source_light_'
images = 'measurements_sersic-sie_sigma_noise=3.9e-03/' #measurements_'
```

```python
deflector_light_data, deflector_mass_data, source_light_data, images_data = [], [], [], []
folder_types = ['Sim1/train/', 'Sim1/test/'] #Sim1: to change in future applications

for sub_folder in folder_types:
    num_sim = len(os.listdir(path + sub_folder + images))
    for i in range(num_sim):
        filename = path + sub_folder + deflector_light + 'lens_light_' + str(i) + '.npy'
        deflector_light_data.append(np.load(filename))
    
        filename = path + sub_folder + deflector_mass + 'lens_mass_' + str(i) + '.npy'
        deflector_mass_data.append(np.load(filename))
    
        filename = path + sub_folder + source_light + 'source_light_' + str(i) + '.npy'
        source_light_data.append(np.load(filename))
    
        filename = path + sub_folder + images+ 'measurements_' + str(i) + '.npy'
        images_data.append(np.load(filename))
```

```python
column_names1 = ['x_lens_light', 'y_lens_light', 'q_lens_light', 'phi_lens_light', 'n_sersic_lens_light', 
                 'R_sersic_lens_light', 'Io_lens']
column_names2 = ['center_x_lens', 'center_y_lens', 'q_lens_mass', 'phi_lens_mass', 'EINSTEIN_RADIUS']
column_names3 = ['center_x_source', 'center_y_source', 'q_source', 'phi_source', 'n_sersic_source_light', 
                 'R_sersic_source_light', 'Io_source']

deflector_light_data = pd.DataFrame(deflector_light_data, columns=column_names1)
deflector_mass_data  = pd.DataFrame(deflector_mass_data, columns=column_names2)
source_light_data    = pd.DataFrame(source_light_data, columns=column_names3)

complete_data = pd.concat([deflector_light_data, deflector_mass_data, source_light_data], axis=1)
complete_images = np.array(images_data)

print(complete_data.shape, complete_images.shape)

complete_data.to_csv('Tianao_Processed/Sim1/data.csv', index=False)
np.save('images.npy', complete_images)
```

```python
import pandas as pd
import numpy as np
import os
import ast
from tqdm.notebook import tqdm

from astropy.visualization import make_lupton_rgb
from astropy.io import fits

import matplotlib.pyplot as plt
plt.style.use('dark_background')
```

```python
#Opening csv:
path = '/Users/jimenagonzalez/research/SkAI_research/SkAI_SLmodeling/Data/Euclid_morphological_catalog/cutouts_vis/'
data = pd.read_csv(path + 'data.csv')
data.head()
```

```python
#Getting images as an array:
images = []

images = []
valid_indices = []
n90, n100 = 0, 0

for i in range(len(data)):
    filename = str(data['object_id'].iloc[i]) + '.fits'
    filepath = path + filename
    if os.path.exists(filepath):
        with fits.open(filepath, memmap=False) as hdul:
            if(hdul[0].data.shape == (99, 99) ):
                n90+=1
                images.append(hdul[0].data)
                valid_indices.append(i)
            else: 
                n100+=1
    else:
        print(f"File not found, skipping: {filename}")

print(n90, n100)
data = data.iloc[valid_indices].reset_index(drop=True)
stacked = np.stack(images)

print(len(data), len(images))
np.save(path + 'images.npy', stacked)
data.to_csv(path + 'data_fixed.csv')
```
